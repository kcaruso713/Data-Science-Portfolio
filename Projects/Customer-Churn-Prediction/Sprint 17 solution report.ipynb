# Final Project Solution Report

This project focused on developing a machine learning model to predict whether an Interconnect customer is likely to churn. The final worklflow included data preparation, exploratory analysis, feature engineering, model training and tuning, and final model evaluation.

## What steps of the plan were performed, and what steps were skipped?

All major steps from the proposed work plan were completed.

The four provided datasets were loaded inspected, and merged using the customer ID. The data was checked for structure, data types, missing values, duplicates, and the distribution of the churn target.

The data was then preprocessed TotalCharges was converted to a numeric variable before encoding, and service-related missing values were handled explicitly. Exploratory data analysis was performed to examine customer characteristics and patterns associated with churn.

Feature engineering was also completed. The churn target was created from the contract end date, and customer tenure was calculated using Feburary 1, 2020 as the observation cutoff date to prevent target leakage. Estimated total charges were recalculated using the corrected tenure value.

The data was split into training, validation, and test sets using stratification, and class imbalance was addressed through class weighting. Several classification models were trained and compared, stronger models were tuned, and the final model was evaluated on the test set.

No major steps from the proposed plan were skipped. Some implementation details were adjusted during the project when issues were identified, especially to prevent target leakage and prepare the data correctly for modeling.

## What difficulties did you encounter and how did you solve them?

One difficulty involved the TotalCharges feature, which initially contained non-numeric values. Encoding the data before correcting this column created unnecessary dimensionality. Thus was resolved by converting TotalCharges to a numeric variable before categorical encoding.

Another important issue was target leakage. An initial tenure calculation used information from contract end dates, which could give the model information that would not relistically be avaliable when predicting churn. To solve this, tenure was recalculated using February 1, 2020 as the cutoff date, and estimated total charges were recalculated using the corrected tenure values.

The churn classes were also imbalanced because fewer customers churned than remained with the company. This was addressed by using stratification when splitting the data and class weighting during model training.

## What were some of the key steps to solving the task?

Some of the most important steps were correctly merging the four datasets, cleaning the data before encoding, creating the churn target, and engineering customer tenure without introducing target leakage.

It was also important to split the data into separate training, validation, and test sets using stratification so that model performance could be evaluated fairly.

Several classification models were compared rather than relying on a single algorithm. The stronger models were tuned, and the model with the best validation performance was selected for final testing.

Keeping the test set separate until the final evalution provided a more reliable estimate of how well the final model should perform on unseen customer data.

## What is your final model, and what quality score does it have?

The final model was the LightGBM Classifier (LGBMClassifier)

On the test dataset, the model achieved:
- ROC-AUC: 0.95
- Accuracy: 0.91

These results indicate that the final model performs strongly at distinguishing customers who are at higher risk of churn from customers who are likely to remain with Interconnect.

Interconnect could use this model to assign churn-risk scores to current customers and identify customers who may benefit from retention efforts, such as promotional offers, discounts, or alternative service plans.
