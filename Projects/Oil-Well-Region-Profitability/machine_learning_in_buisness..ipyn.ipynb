```python
#Variables for profit calculation
budget = 100_000_000
revenue_per_unit = 4500
wells_to_select = 200
```

```python
def calculate_profit(target, predictions, count):
    predictions_sorted = predictions.sort_values(ascending=False)
    selected = target[predictions_sorted.index][:count]

    revenue = selected.sum() * revenue_per_unit
    profit = revenue - budget

    return profit
```

<b>   Reviewer's comment </b>

Correct! We select the best 200 wells by predictions, but then we need to use target values to actually calculate profit because predictions may not be accurate.

```python
#Calculate profit for each region

predictions_valid_0 = pd.Series(predictions_valid_0, index=target_valid_0.index)
predictions_valid_1 = pd.Series(predictions_valid_1, index=target_valid_1.index)
predictions_valid_2 = pd.Series(predictions_valid_2, index=target_valid_2.index)

profit_0 = calculate_profit(target_valid_0, predictions_valid_0, wells_to_select)
profit_1 = calculate_profit(target_valid_1, predictions_valid_1, wells_to_select)
profit_2 = calculate_profit(target_valid_2, predictions_valid_2, wells_to_select)

print("Region 0 profit:", profit_0)
print("Region 1 profit:", profit_1)
print("Region 2 profit:", profit_2)
```

```python
##Profit Calculation Conclusions
#Based on the 200 wells with the highest predicted reserves, all three regions would generate a profit.
#Region 0 has the highest estimated profit at approximately $33.21 million, followed by Region 2 at approximately $27.10 million and Region 1 at approximately $24.15 million
#However, these results alone are not enough to select a region because the risk of losses must also be evaluated using bootstrapping.
```

```python
##Bootstrapping and Risk Analysis
#Bootstrapping with 1000 samples is used to estimate the distribution of potential profit for each region. For each bootstrap sample, 500 wells are sampled with replacement , and the 200 wells with the highest predicted reserves are selected for profit calculation.
#The average profit, 95% confidence interval, and risk of loss are calculated for each region. A region is considered acceptable only if its risk of loss is below 2.5%
```

```python
def bootstrap_profit(target, predictions):
    state = np.random.RandomState(12345)
    values = []

    for i in range(1000):
        target_subsample = target.sample(
            n=500,
            replace=True,
            random_state=state
        )

        predictions_subsample = predictions[target_subsample.index]

        profit = calculate_profit(
            target_subsample,
            predictions_subsample,
            wells_to_select
        )

        values.append(profit)

    values = pd.Series(values)

    mean_profit = values.mean()
    lower = values.quantile(0.025)
    upper = values.quantile(0.975)
    risk_of_loss = (values < 0 ).mean() * 100

    return mean_profit, lower, upper, risk_of_loss
```

```python
mean_profit_0, lower_0, upper_0, risk_of_loss_0 = bootstrap_profit(
    target_valid_0, predictions_valid_0
)
mean_profit_1, lower_1, upper_1, risk_of_loss_1 = bootstrap_profit(
    target_valid_1, predictions_valid_1
)
mean_profit_2, lower_2, upper_2, risk_of_loss_2 = bootstrap_profit(
    target_valid_2, predictions_valid_2
)

print("Region 0 average profit:", mean_profit_0)
print("Region 0 95% confidence interval:", lower_0, upper_0)
print("Region 0 risk of loss:", risk_of_loss_0, "%")

print()

print("Region 1 average profit:", mean_profit_1)
print("Region 1 95% confidence interval:", lower_1, upper_1)
print("Region 1 risk of loss:", risk_of_loss_1, "%")

print()

print("Region 2 average profit:", mean_profit_2)
print("Region 2 95% confidence interval:", lower_2, upper_2)
print("Region 2 risk of loss:", risk_of_loss_2, "%")
```
