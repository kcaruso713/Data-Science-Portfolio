```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Import the files
company_data = pd.read_csv('/datasets/project_sql_result_01.csv')
neighborhood_data = pd.read_csv('/datasets/project_sql_result_04.csv')

# 2. Study the data and check data types
display(company_data.head())
display(neighborhood_data.head())


print(company_data.info())
print(neighborhood_data.info())

# 3. Identify the top 10 neighborhoods by drop-offs
top_10_neighborhoods = neighborhood_data.sort_values(by='average_trips', ascending=False).head(10)
print(top_10_neighborhoods)

# 4. Make Graphs
# Graph 1: Top 10 Taxi Companies and number of rides
plt.figure(figsize=(12, 6))
sns.barplot(data=company_data.sort_values(by='trips_amount', ascending=False).head(10), 
            x='trips_amount', y='company_name')
plt.title('Top 10 Taxi Companies by Number of Rides (Nov 15-16, 2017)')
plt.show()
#Flash Cab had the highest number of rides on November 15-16, 2017. The top companies had much higher ride counts than the rest, showing that rides are concentrated among a few major taxi companies.

# Graph 2: Top 10 Neighborhoods by number of drop-offs
plt.figure(figsize=(12, 6))
sns.barplot(data=top_10_neighborhoods, x='average_trips', y='dropoff_location_name')
plt.title('Top 10 Neighborhoods by Average Drop-offs (Nov 2017)')
plt.show()

```

```python
##Conclusion
##The Loop had the highest average number of drop-offs, followed by River North and Streeterville. These are central/popular areas, so it makes sense they receive more taxi drop-offs.
```

```python
from scipy import stats as st

# Import the third dataset
loop_ohare_trips = pd.read_csv('/datasets/project_sql_result_07.csv')

display(loop_ohare_trips.head())
loop_ohare_trips.info()
# Segregate the data into two groups
good_weather = loop_ohare_trips[loop_ohare_trips['weather_conditions'] == 'Good']['duration_seconds']
bad_weather = loop_ohare_trips[loop_ohare_trips['weather_conditions'] == 'Bad']['duration_seconds']

# Set significance level
alpha = 0.05

# Perform T-Test (Independent samples)
results = st.ttest_ind(good_weather, bad_weather, equal_var=False)

print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print("We reject the null hypothesis.")
else:
    print("We cannot reject the null hypothesis.")
```

```python
## Conclusion
#Null Hypothesis: The average duration from the loop to O'Hare is the same on rainy saturdays and non-rainy saturdays.t-test 
#Alternative hypothesis: The average duration change on rainy Saturdays.
#I chose alpha = 0.05 because it is a common significance level for hypothesis testing.
#I used an independent two-sample because I am comparing the average durationof two independent groups: rides in good weather and rides in bad/rainy weather.



#Since the p-value is less than 0.05, wereject the null hypothesis. There is statistically significant evidence that the average duration of rides from the Loop to O'Hare changes on rainy Saturdays.
```
